## Comandos del sistemas de archivos
Al ingresar, todo usuario se ubica en su directorio hogar (HOME), para saber cuál es la ubicación en la que se encuentra en el sistema de archivos o su directorio de trabajo, se usa el comando **pwd** (print work  directory).
```bash
pwd
```
Para conocer el contenido de un directorio se usa el comando **ls** (list). Tiene opciones como *-a* que muestra los archivos ocultos o *-l* que muestra más detalles (permisos, accesos directos, autor, grupo, tamaño, fecha de modificaión) de los directorios.
```bash
ls
ls -l
```
Todos los comandos cuentan con una ayuda básica que puede obtenerse mediante la opción *--help* del comando o mediante el comando **man** (manual) seguido del nombre del comando a consultar.
```bash
pwd --help
```
```bash
man pwd
```

````python
%%bash
pwd
ls
ls -l
man pwd
````

Para crear directorios se usa el comando **mkdir** (make directory) seguido del nombre del directorio o directorios a crear.

El comando **cp** (copy) permite copiar archivos (la opcion *-r* copia todas las subcarpetas), el **mv** (move) los mueve o renombra y sus sintáxis son:
```bash
cp origen destino
mv origen destino
```
Los comandos **rm** y **rmdir** (remove) se utilizan para borrar uno o varios archivos y directorios respectivamente, donde rmdir requiere que el directorio se encuentre vacío. La opcion *rm -r* hace recursivo el remove.

Con **tree** se obtiene una visualización en forma de ramas.

````python
%%bash
pwd
tree

mkdir ../BORRAME
tree -L 1 ../

mv ../BORRAME ./
tree

rmdir BORRAME/
tree
````

Los comandos **cat**, **more** y **less** permiten ver el contenido de un archivo de texto

- cat: vacía el contenido de un archivo en la salida estándar (pantalla) desde el inicio hasta el final sin interrupción.
- more: muestra el contenido de un archivo por páginas que se pueden avanzar mediante la barra espaciadora o con la teclar ENTER línea por línea.
- less: muestra el contenido del archivo en forma dinámica, permite desplazarse a lo largo del archivo hacia arriba o abajo mediante las flechas.

Los comandos **head** y **tail** permiten ver el inicio o el final de un archivo, por defecto muestras 10 lineas pero se puede cambiar con "-" seguido del numero de líneas a desplegar.

Para imprimir un numero aleatorio del 0 al 99:

Declarar una variable y ver su contenido/valor:

Regresa el resultado de una operacion o cadena sencilla unicamente con enteros (cuidado con los espacios):
```bash
expr a + 1 \* 2
```
```bash
a=`expr $a + 1`
```

````python
%%bash
rndm=$(($RANDOM%100))
echo $rndm
a=10
echo $a
a=`expr $a + 100`
echo $a
b=`expr $a \* 2 + $rndm`
echo $b
c=`expr \( $a - 10 \) / 2`
echo $c
````

### Usuarios, permisos y tipos de archivos
Unix maneja 3 niveles para el control de acceso a los archivos:

- Nivel de *usuario*: el propietario original de los archivos tiene todo el control sobre sus archivos.
- Nivel de *grupo*: todos los miembros de un grupo tienen acceso a los archivos compartidos por el grupo, pero el control le pertenece al propietario original del archivo.
- *Otros*: esto aplica al resto de los usuarios del sistema, el control de acceso lo determina el propietario del sistema.

Los permisos de acceso a un archivo son de tres tipos:

- **Lectura (r)**: el archivo se puede leer
- **Escritura (w)**: el archivo puede ser modificado, borrado o creado
- **Ejecución (x)**: el archivo puede ejecutarse, esto funciona con programas binarios ejecutables o scripts de algún intérprete, incluso el mismo intérprete de comandos.

**Forma octal**: los tres bits de un dígito octal corresponde a los permisos rwx de tal forma que para activar un permiso se convierte el permiso deseado en su equivalente en octal. Se usan 3 dígitos octales para cada nivel de seguridad (usuario, grupo, otros), donde:
- 1 = x
- 2 = w
- 4 = r

````python
%%bash
touch archivo
ls -l archivo
chmod 000 archivo
ls -l archivo
chmod 700 archivo
ls -l archivo
chmod 067 archivo
ls -l archivo
chmod 050 archivo
ls -l archivo
rm archivo
````

### Estructuras de control de flujo (for, if)
- FOR
```bash
for variable in elemento1 ... elementoN
do
    # Comandos que usan $variable
    echo $variable
done
```

- IF
```bash
if [ condicion_1 ]; then
    # Se ejecuta si condicion_1 es verdadera
elif [ condicion_2 ]; then
    # Se ejecuta si condicion_1 es falsa y condicion_2 verdadera
else
    # Se ejecuta si todas las condiciones anteriores son falsas
fi
```
donde las condiciones se manejan con operadores de comparacion:\
- Numèricos: -eq (igual), -ne (distinto), -lt (menor que), -gt (mayor que), -le (menor o igual), -ge (mayor o igual).
- Cadenas: = (igual), != (distinto), -z (cadena vacía).
- Archvivos: -f (es archivo), -d (es directorio), -r (legible), -e (existe).

````python
%%bash
for i in {1..5}; do echo "Número: $i"; done

a=34
b=34
if [ "$a" -eq "$b" ]; then
    echo "a y b son iguales"
fi
````
