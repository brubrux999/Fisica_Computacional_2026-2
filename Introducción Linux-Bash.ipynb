### Comandos del sistemas de archivos
Al ingresar, todo usuario se ubica en su directorio hogar (HOME), para saber cuál es la ubicación en la que se encuentra en el sistema de archivos o su directorio de trabajo, se usa el comando **pwd** (print work  directory).
```bash
pwd
```
Para conocer el contenido de un directorio se usa el comando **ls** (list). Tiene opciones como *-a* que muestra los archivos ocultos o *-l* que muestra más detalles (permisos, accesos directos, autor, grupo, tamaño, fecha de modificaión) de los directorios.
```bash
ls
ls -l
```
Todos los comandos cuentan con una ayuda básica que puede obtenerse mediante la opción *--help* del comando o mediante el comando **man** (manual) seguido del nombre del comando a consultar.
```bash
pwd --help
```
```bash
man pwd
```

````python
%%bash
pwd
ls
ls -l
man pwd
````

Para desplazarse en el sistema de archivos se utiliza el comando **cd** (change directory) seguido del *PATH*. Existen 2 directorios especiales; "." hace referencia al directorio actual y ".." hace referencia al directorio superior (hacia la raíz)

````python
%%bash
cd .
pwd
cd ..
pwd
cd Fisica_Computacional_2026-2/
pwd
cd /mnt/c/Users/bruny/Desktop/
pwd
cd
pwd
````
